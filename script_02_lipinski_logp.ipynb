```python
from rdkit import Chem
from rdkit.Chem import Descriptors
```

```python
#Define a target molecule (In this scenario I picked Quercetin, a natural bioactive molecule)

compound_name="Quercetin"
smiles_code="O=C1c3c(O/C(=C1/O)c2ccc(O)c(O)c2)cc(O)cc3O"

molecule=Chem.MolFromSmiles(smiles_code)
```

```python
#Calculating two parameters from the Lipinski's rule of 5
weight=round(Descriptors.MolWt(molecule), 2)
logp= round(Descriptors.MolLogP(molecule),2)
```

```python
print(f"Molecule Analysis: {compound_name}")
print("-" * 50)
print(f"1. Molecular Weight: {weight} Da (Threshold < 500)")
print(f"2. Lipophilicity (LogP): {logp} (Threshold < 5)")

if weight < 500 and logp < 5:
    print("PASS. The molecule fulfills the criteria.")
else:
    print("FAIL. The molecule violates the criteria")
```
