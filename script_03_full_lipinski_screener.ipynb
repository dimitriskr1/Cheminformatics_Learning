```python
from rdkit import Chem
from rdkit.Chem import Descriptors
```

```python
#Creating the mini libraby

compound_library = {
    "Aspirin" : "O=C(C)Oc1ccccc1C(=O)O" ,
    "Quercetin" : "O=C1c3c(O/C(=C1/O)c2ccc(O)c(O)c2)cc(O)cc3O" ,
    "Erythromycin" : "CC[C@H]1OC(=O)[C@H](C)[C@@H](O[C@H]2C[C@@](C)(OC)[C@@H](O)[C@H](C)O2)[C@H](C)[C@@H](O[C@@H]2O[C@H](C)C[C@@H]([C@H]2O)N(C)C)[C@](C)(O)C[C@@H](C)C(=O)[C@H](C)[C@@H](O)[C@]1(C)O"

}

print("--- Lipinski RO5 Screener ---")
```

```python
for name, smiles in compound_library.items():
    mol = Chem.MolFromSmiles(smiles)

    #Calculating 4 of the Lipinski's parameters
    mw = Descriptors.MolWt(mol)
    logp = Descriptors.MolLogP(mol)
    h_donors = Descriptors.NumHDonors(mol)
    h_acceptors = Descriptors.NumHAcceptors(mol)


    # Counting the violations (maximum violations : 1)
    violations = 0 

    if mw > 500:
        violations += 1
    if logp > 5:
        violations += 1
    if h_donors > 5:
        violations += 1
    if h_acceptors > 10:
        violations += 1


    #Displaying the results for the current examined molecule
    print(f"\nAnalyzing: {name}")
    print(f"MW: {round(mw,2)} | LogP: {round(logp,2)} | H-Donors: {h_donors} | H-Acceptors: {h_acceptors}")
    print(f"Total Violations: {violations}")

# 4. Final Decision making
    if violations <= 1:
        print("Result: PASSES (Drug-like compound)")
    else:
        print("Result: FAILS (Too many Lipinski violations)")
```
