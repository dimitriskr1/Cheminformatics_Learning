```python
import pandas as pd
from rdkit import Chem
from rdkit.Chem import Descriptors
```

```python
compound_library = {
    "Aspirin" : "O=C(C)Oc1ccccc1C(=O)O" ,
        "Quercetin" : "O=C1c3c(O/C(=C1/O)c2ccc(O)c(O)c2)cc(O)cc3O" ,
        "Erythromycin" : "CC[C@H]1OC(=O)[C@H](C)[C@@H](O[C@H]2C[C@@](C)(OC)[C@@H](O)[C@H](C)O2)[C@H](C)[C@@H](O[C@@H]2O[C@H](C)C[C@@H]([C@H]2O)N(C)C)[C@](C)(O)C[C@@H](C)C(=O)[C@H](C)[C@@H](O)[C@]1(C)O" 
}
```

```python
# Creating an empty list to store our results
data_results = []

print("Running Screening...")

for name, smiles in compound_library.items():
    mol = Chem.MolFromSmiles(smiles)

    mw=Descriptors.MolWt(mol)
    logp=Descriptors.MolLogP(mol)
    h_donors=Descriptors.NumHDonors(mol)
    h_acceptors=Descriptors.NumHAcceptors(mol)

    violations = 0
    if mw > 500: violations += 1
    if logp > 5: violations += 1
    if h_donors > 5: violations += 1
    if h_acceptors > 10: violations += 1

    status = "Pass" if violations == 0 else "Fail"


    # Appending the results to our list
    data_results.append({
        "Compound Name": name,
        "Molecular Weight": mw,
        "LogP": logp,
        "H-Bond Donors": h_donors,
        "H-Bond Acceptors": h_acceptors,
        "Violations": violations,
        "Lipinski Status": status
    })
```

```python
# Converting the list of results into a DataFrame
df = pd.DataFrame(data_results)
```

```python
# Exporting the DataFrame to a CSV file
csv_filename="lipinski_report.csv"
df.to_csv(csv_filename, index=False)
print(f"Screening complete. Results saved to {csv_filename}.\n")
```
