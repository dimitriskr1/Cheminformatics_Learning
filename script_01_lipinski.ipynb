```python
# Import basic cheminfo tools
from rdkit import Chem
from rdkit.Chem import Descriptors

# 1. Variables: Naming a name and a code to the examined molecule
drug_name = "Aspirin"
smiles_code = "CC(=O)Oc1ccccc1C(=O)O"

# 2. This steps calculates the molecular weight of the molecule
molecule = Chem.MolFromSmiles(smiles_code)
weight = Descriptors.MolWt(molecule)

# Printing the results
print("Drug:", drug_name)
print("Molecular Weight:", round(weight, 2), "g/mol")
print("-----------------------------------")

# 3. Conditional Logic (based on the Lipinski Rule of 5 for the MW criteria)
if weight < 500:
    print("Result: The molecule passes the test.")
else:
    print("Result: The molecule failed the test. The MW surpasses Lipinski's MW criteria.")
```
